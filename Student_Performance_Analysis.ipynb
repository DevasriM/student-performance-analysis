# Student Performance Analysis

## Project Overview
This project explores student academic performance using Python, Pandas, NumPy, Matplotlib, and Seaborn. It analyzes student grades and selected factors such as study time, previous failures, and absences to identify patterns in academic performance.

## Objectives
- Analyze students' final grades using descriptive statistics.
- Examine the relationship between study time and final grades.
- Compare academic performance across previous-failure categories.
- Investigate the relationship between absences and final grades.
- Explore correlations between selected academic variables.
- Clean and prepare the dataset for reliable analysis.

## Dataset
The dataset contains 395 student records and 33 attributes, including demographic information, family and school-related factors, study time, absences, previous failures, and three period grades: G1, G2, and G3.

The final grade is represented by **G3**, with a scale from 0 to 20.

## Tools and Technologies
- Python
- Pandas and NumPy
- Matplotlib and Seaborn
- Google Colab
- Excel/CSV data handling

## Analysis Workflow
1. Load the dataset.
2. Correct the semicolon-separated data format.
3. Inspect column names, missing values, and duplicate records.
4. Clean the data and convert relevant columns to numeric types.
5. Calculate descriptive statistics and performance measures.
6. Analyze study time, previous failures, and absences.
7. Examine correlations and summarize findings.

## Key Findings
- Number of student records analyzed: 395.
- Average final grade: 10.42/20.
- Median final grade: 11/20.
- Students scoring 10 or above: 265 (67.09%).
- The study-time category with the highest observed average final grade was category 3.
- Students with more previous failures had lower average final grades in this dataset.
- The correlation between absences and final grades was approximately 0.034, indicating a very weak linear relationship.

## Conclusion
The analysis identifies patterns and associations in student performance. In this dataset, previous failures show a clearer negative association with final grades than absences do. These findings are descriptive and do not establish cause and effect.

## Limitations
The findings are based on one dataset and should not be generalized to all students. Correlation does not imply causation. The study-time categories represent coded groups, not exact hours.

Open the notebook in Google Colab, upload the required dataset when prompted, and run the cells from top to bottom.

```python

# ============================================================
# STUDENT PERFORMANCE ANALYSIS
# Python | Pandas | NumPy | Matplotlib | Seaborn
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
from IPython.display import display
import io

print("STUDENT PERFORMANCE ANALYSIS")
print("=" * 50)

# 1. UPLOAD EXCEL FILE FROM YOUR PC
print("\nPlease upload student_mat.xlsx from your PC.")
uploaded = files.upload()

if not uploaded:
    raise ValueError("No file uploaded. Please run this cell again.")

filename = next(iter(uploaded))
print(f"\nUploaded file: {filename}")

# 2. READ THE EXCEL FILE
raw = pd.read_excel(io.BytesIO(uploaded[filename]), header=None)

# 3. FIX SEMICOLON-SEPARATED DATA
# If the Excel sheet contains the entire CSV in one column,
# split that column into the actual 33 columns.
if raw.shape[1] == 1:
    first_value = str(raw.iloc[0, 0])

    if ";" in first_value:
        print("Semicolon-separated data detected. Splitting columns...")
        df = raw.iloc[:, 0].astype(str).str.split(";", expand=True)

        # First row contains the actual column headers
        df.columns = (
            df.iloc[0]
            .astype(str)
            .str.strip()
            .str.strip('"')
            .str.lower()
        )

        # Remove header row
        df = df.iloc[1:].reset_index(drop=True)
    else:
        raise ValueError(
            "The Excel file has one column but no semicolon separator. "
            "Please check the uploaded file."
        )
else:
    # The Excel sheet already has separate columns
    df = raw.copy()

    # If the first row contains column names, use it as the header
    df.columns = (
        df.iloc[0]
        .astype(str)
        .str.strip()
        .str.strip('"')
        .str.lower()
    )
    df = df.iloc[1:].reset_index(drop=True)

# 4. STANDARDIZE ALL VALUES
# Clean text values safely, one column (Series) at a time.
for col in df.columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.strip()
        .str.strip('"')
    )

df.columns = (
    df.columns.astype(str).str.strip().str.lower()
)

# Convert numeric columns
numeric_columns = [
    "age", "medu", "fedu", "traveltime", "studytime",
    "failures", "famrel", "freetime", "goout",
    "dalc", "walc", "health", "absences", "g1", "g2", "g3"
]

for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# 5. VALIDATE DATA
required_columns = [
    "studytime", "absences", "failures", "g1", "g2", "g3"
]

missing_columns = [
    col for col in required_columns if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}\n"
        f"Detected columns: {df.columns.tolist()}"
    )

print("\nDataset loaded and corrected!")
print("Dataset dimensions:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst five records:")
display(df.head())

# Stop if data could not be parsed correctly
if df.empty:
    raise ValueError("The dataset is empty after parsing.")

if df["g3"].isna().all():
    raise ValueError(
        "Final grades could not be converted to numbers. "
        "Please check the uploaded file format."
    )

# 6. DATA QUALITY CHECK
print("\n========== DATA QUALITY ==========")
print("Missing values before cleaning:", int(df.isna().sum().sum()))
print("Duplicate rows:", int(df.duplicated().sum()))

df = df.drop_duplicates().copy()
df = df.dropna(subset=required_columns).copy()

print("Rows after cleaning:", len(df))

if df.empty:
    raise ValueError(
        "No valid records remain after cleaning. "
        "Check the numeric columns in the uploaded Excel file."
    )

# 7. OVERALL PERFORMANCE
print("\n========== OVERALL PERFORMANCE ==========")
print("Number of students:", len(df))
print("Average final grade:", round(df["g3"].mean(), 2), "/20")
print("Median final grade:", round(df["g3"].median(), 2), "/20")
print("Highest final grade:", df["g3"].max())
print("Lowest final grade:", df["g3"].min())
print("Students scoring 10 or above:", int((df["g3"] >= 10).sum()))
print(
    "Pass-threshold percentage (grade >= 10):",
    round((df["g3"] >= 10).mean() * 100, 2), "%"
)

print("\nDescriptive statistics:")
display(df.describe())

sns.set_theme(style="whitegrid")

# 8. STUDY TIME ANALYSIS
print("\n========== STUDY TIME ANALYSIS ==========")
study_results = df.groupby("studytime")["g3"].agg(
    ["count", "mean", "median"]
).round(2)
display(study_results)

plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="studytime", y="g3", errorbar=None)
plt.title("Study Time vs Average Final Grade")
plt.xlabel("Study Time Category (1 = Lowest, 4 = Highest)")
plt.ylabel("Average Final Grade")
plt.tight_layout()
plt.show()

# 9. ABSENCES ANALYSIS
print("\n========== ABSENCES ANALYSIS ==========")
absence_corr = df["absences"].corr(df["g3"])
print("Correlation between absences and final grade:",
      round(absence_corr, 3))

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="absences", y="g3", alpha=0.7)
plt.title("Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade")
plt.tight_layout()
plt.show()

# 10. PREVIOUS FAILURES ANALYSIS
print("\n========== PREVIOUS FAILURES ANALYSIS ==========")
failure_results = df.groupby("failures")["g3"].agg(
    ["count", "mean", "median"]
).round(2)
display(failure_results)

plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="failures", y="g3", errorbar=None)
plt.title("Previous Failures vs Average Final Grade")
plt.xlabel("Number of Previous Failures")
plt.ylabel("Average Final Grade")
plt.tight_layout()
plt.show()

# 11. FINAL GRADE DISTRIBUTION
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="g3", bins=15, kde=True)
plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()

# 12. CORRELATION HEATMAP
print("\n========== CORRELATION ANALYSIS ==========")
correlation = df[
    ["studytime", "absences", "failures", "g1", "g2", "g3"]
].corr()

display(correlation.round(2))

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)
plt.title("Correlation Between Academic Variables")
plt.tight_layout()
plt.show()

# 13. AUTOMATIC FINDINGS
print("\n========== PROJECT FINDINGS ==========")

study_means = df.groupby("studytime")["g3"].mean()
failure_means = df.groupby("failures")["g3"].mean()

print(f"Average final grade: {df['g3'].mean():.2f}/20")
print("Study-time category with highest average:",
      int(study_means.idxmax()))
print("Failure category with highest average:",
      int(failure_means.idxmax()))
print(f"Absence-grade correlation: {absence_corr:.3f}")

print(
    "\nImportant: These are observed associations, "
    "not proof of cause and effect."
)

# 14. EXPORT CLEANED DATASET
output_file = "cleaned_student_performance.csv"
df.to_csv(output_file, index=False)

files.download(output_file)

print("\nCleaned dataset downloaded successfully!")
print("STUDENT PERFORMANCE ANALYSIS COMPLETED!")
```
